# Four-task real causal-agent evaluation

Upload the archive produced by `package_real_four_causal_agent_colab_task.py`. The notebook runs the real Qwen model path on the available GPU and writes a digest-checked summary.

```python
from google.colab import files
uploaded = files.upload()
archive_name = next((name for name in uploaded if name.endswith('.tgz')), None)
if archive_name is None:
    raise ValueError('Upload the .tgz archive created by the package script')
with open('/content/real-four-causal-agent.tgz', 'wb') as dst:
    dst.write(uploaded[archive_name])
print('uploaded', archive_name)
```

```python
import os, subprocess
os.environ.pop('REAL_FOUR_CAUSAL_SKIP_MODEL_SETUP', None)
result = subprocess.run([
    'python3', '/content/real-four-causal-agent/colab/run_real_four_causal_agent_remote.py'
], text=True, capture_output=True)
print(result.stdout[-4000:])
print(result.stderr[-4000:])
if result.returncode:
    raise RuntimeError(f'four-task run failed with exit code {result.returncode}')
```

```python
import json
summary_path = '/content/real-four-causal-agent/artifacts/real-four-causal-agent-colab-summary.json'
summary = json.load(open(summary_path))
print(json.dumps({'status': summary['status'], 'metrics': summary['metrics']}, indent=2))
files.download(summary_path)
```
